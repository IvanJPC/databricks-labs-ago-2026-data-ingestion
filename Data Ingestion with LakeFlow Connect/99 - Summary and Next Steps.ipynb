##### EXPAND FOR ADDITIONAL NOTES

<details>

#### The Complete LakeFlow Connect Toolkit

This visual shows the three pillars of data ingestion you explored across all lectures, demos, and labs. Each builds on the others to deliver governed, analysis-ready bronze tables in the Lakehouse.

#### LakeFlow Connect Connectors

LakeFlow Connect is the ingestion layer of the Databricks Data Intelligence Platform &mdash; it replaces the traditional patchwork of ingestion tools with a single managed surface:

- **Upload Files** &mdash; ad-hoc loads of local files into Unity Catalog volumes or tables
- **Standard Connectors** &mdash; cloud object storage, Kafka, and other sources, supporting batch, incremental batch, and streaming ingestion
- **Managed Connectors** &mdash; fully managed, UI-driven ingestion from enterprise databases and SaaS applications (Salesforce, Workday, SQL Server, and more)
- **Partner Connect** &mdash; validated partner solutions when a native managed connector isn't available

#### Ingestion Methods

The right method depends on data volume, freshness, and operational pattern:

- **CREATE TABLE AS (CTAS)** with `read_files()` &mdash; simple batch ingestion that creates Delta tables from raw files. Best for one-time or smaller ad-hoc loads.
- **COPY INTO** &mdash; idempotent, retriable incremental batch ingestion. Skips files already loaded; ideal for scheduled pipelines handling thousands of files.
- **Auto Loader** (Python `cloudFiles` or SQL streaming tables in Declarative Pipelines) &mdash; the most scalable option. Handles automatic schema evolution and processes billions of files for incremental or streaming workloads.
- **MERGE INTO** &mdash; atomic upserts into existing Delta tables. The right tool for slowly changing dimensions, incremental loads, and change data capture.

#### Bronze Quality

Production bronze tables need more than raw values. Throughout the course you enriched and protected your ingested data:

- **`_metadata` columns** &mdash; capture `file_name` and `file_modification_time` during ingestion for lineage, auditing, and debugging
- **`_rescued_data` column** &mdash; preserve records that don't match the target schema as JSON-formatted strings, preventing silent data loss
- **JSON handling** &mdash; flatten semi-structured data into queryable columns using STRING, STRUCT (with `schema_of_json` + `from_json`), or the new VARIANT data type

#### The Ingest & Refine Loop

The coral bar represents the continuous flow of data through the Medallion Architecture. Bronze tables produced by LakeFlow Connect feed cleaned Silver tables, which feed curated Gold tables &mdash; with each layer refined incrementally and governed by Unity Catalog.

</details>
