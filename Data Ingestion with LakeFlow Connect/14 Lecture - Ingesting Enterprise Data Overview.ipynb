# Lecture - Ingesting Enterprise Data Overview

## Overview

In this lecture, you will learn how Lakeflow Connect Managed Connectors and Partner Connect streamline enterprise data ingestion by enabling fast and reliable integration from databases and applications into the Databricks Lakehouse through flexible, fully managed, and partner-supported options.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Explain the need for managed connectors** when ingesting data from enterprise databases and SaaS applications beyond cloud object storage
2. **Describe the benefits of LakeFlow Connect Managed Connectors** including simplified setup, UI-driven configuration, and fully managed infrastructure
3. **Compare the SaaS and database ingestion architectures** used by LakeFlow Connect Managed Connectors
4. **Describe Partner Connect** as an alternative for data sources without a native managed connector

## D. Lakeflow Connect Managed Connectors: SaaS Ingestion 

Let's start with an overview of the Lakeflow Connect managed connector architecture for SaaS applications.

Lakeflow Connect enables data ingestion from external, publicly accessible sources such as APIs or OLAP endpoints into Streaming Delta Tables, using serverless, declarative pipelines. You can setup these pipelines using the user interface (UI) or the API. Managed connectors leverage efficient incremental reads and writes to make data ingestion faster, scalable, and more cost-efficient, while your data remains fresh for downstream consumption

Finally, for SaaS connectors, all data movement happens in the data plane. The control plane is only used for pipeline setup, monitoring (e.g., reading event logs), and management.

## E. Database Ingestion Architecture

Now let's look at how the architecture changes when using a LakeFlow Connect managed database connector.

Like with SaaS connectors, this architecture is designed to move data into Streaming Delta Tables — but this time from external databases rather than public APIs.

  <div style="font-size: 14pt; margin-bottom: 10px;">LakeFlow Connect collects data from <strong>external databases</strong> to Streaming Delta Tables:</div>
  <ol style="font-size: 13pt; line-height: 1.8; padding-left: 22px; margin: 0;">
    <li>The classic compute Declarative Pipelines job <strong>collects credentials</strong> from UC</li>
    <li>It uses the credentials to <strong>connect and collect data</strong> from your Database sources</li>
    <li>The latest <strong>state and staging data are saved</strong> to your Unity Catalog volume</li>
    <li>A Serverless Declarative Pipelines job <strong>processes the collected data</strong> to your Streaming Delta Tables</li>
  </ol>

##### EXPAND FOR ADDITIONAL NOTES
<details>

The database ingestion architecture is more involved than SaaS because it needs to connect to databases that may be on-prem or in a private cloud.

<strong>Key Components Introduced</strong>

We're introducing two new architectural elements here:
<ol>
  <li><strong>Ingestion Gateway:</strong>
    <ul>
      <li>A dedicated pipeline that connects to the database to extract:
        <ul>
          <li>Metadata</li>
          <li>Snapshots</li>
          <li>Change logs — It stages all of this in the Unity Catalog (UC) volume.</li>
        </ul>
      </li>
    </ul>
  </li>
  <li><strong>Unity Catalog Volume:</strong>
    <ul>
      <li>This acts as the intermediate staging layer, enabling the next pipeline to pick up and stream data.</li>
      <li>It's secured using standard UC mechanics, and by default, access is limited to the user running the pipeline.</li>
    </ul>
  </li>
</ol>

So why did we even add this gateway step?

For starters, it helps with networking. Many customers have databases that are (1) sitting behind a firewall, (2) not publicly accessible to the internet, and so on. If you don't have an option like Private Link, we can deploy the gateway inside of your network.

The gateway also helps us limit the load that we're placing on the database. After all, you probably want to limit the number of direct connections to your data source. By splitting out the gateway from the pipeline, we'll be able to have one gateway speak to the database — and then fan out to N pipelines for that scalability. (For context, this isn't a problem in a SaaS connector because the load is typically controlled with API limits. But even there, we're using those limits as efficiently as we can.)

</details>

## G. Conclusion

In this lecture, you learned how to ingest enterprise data into Databricks beyond cloud object storage:

- **LakeFlow Connect Managed Connectors** simplify ingestion from enterprise databases and SaaS applications with a fully managed, UI-driven or API-driven experience.
- **SaaS ingestion architecture**: A serverless Declarative Pipelines job collects credentials from Unity Catalog, reaches out to the public data source, and writes to Streaming Delta Tables.
- **Database ingestion architecture**: Uses an additional Ingestion Gateway (classic compute) to connect to private databases, with staging and state management in Unity Catalog volumes, before a serverless pipeline writes to Streaming Delta Tables.
- **Partner Connect** provides an alternative when no native managed connector is available, offering a rich ecosystem of partner solutions (Informatica, Qlik, Rivery, Alteryx, Prophecy, Fivetran, and more).
