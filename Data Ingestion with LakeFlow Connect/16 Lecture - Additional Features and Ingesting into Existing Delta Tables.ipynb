# Lecture - Additional Features and Ingesting into Existing Delta Tables

## Overview

Databricks provides additional features such as Lakehouse Federation, Zerobus, Delta Sharing, and Databricks Marketplace to expand data integration, sharing, and collaboration capabilities in the Lakehouse.

This lecture also introduces how `MERGE INTO` streamlines ingestion into existing Delta tables by applying updates, inserts, and deletes from a source in a single atomic operation.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Describe additional Databricks features** that extend data integration, sharing, and collaboration capabilities
2. **Identify key Databricks Marketplace components** and explore shared data assets
3. **Explain the purpose of MERGE INTO** for applying updates, inserts, and deletes to existing Delta tables
4. **Describe MERGE INTO clauses** including matched updates, matched deletes, and not matched inserts
5. **Write a MERGE INTO statement** to merge source data into a target Delta table

## D. Updates, Inserts, and Deletes on Delta Tables with MERGE INTO

The MERGE INTO command allows you to apply updates, inserts, and deletes from a source table into an existing Delta table, all in a single, atomic operation.


### D1. MERGE INTO Overview

<ol>
  <li><strong>Step 1: Declare Target and Source Table</strong></br>
  <ul>
  <li>MERGE INTO <code>target_table</code> target - This declares the target table, the table you want to update, insert into, or delete from. You're giving it the alias target to reference it later in the query.</li>
  <li>USING <code>source_table</code> source - This defines the source table, the table that contains the new data or changes you want to apply to the target. It's given the alias source.</li>
  </ul>
  </li>

  <br>
  <li><strong>Step 2: Specify the Merge Condition</strong><br>
  <code>ON target.id = source.id</code> defines the matching condition between the source and target tables, typically using a unique key like <code>id</code>.
  </li>

  <br>
  <li><strong>Step 3: WHEN MATCHED - UPDATE</strong><br>
  WHEN MATCHED AND source.status = 'update' THEN

  If a row exists in both tables and the source indicates it's an update.

  UPDATE SET  - Updates the target table with new values from the source:
  <ul>
  <li>email</li>
  <li>status</li>
  </ul>
  </li>
  <br>
  <li><strong>Step 4: WHEN MATCHED - DELETE</strong><br>
  WHEN MATCHED AND source.status = 'delete' THEN DELETE

  If the row exists in both tables and the source says it should be deleted, it’s removed from the target.

  <br>
  <li><strong>Step 5: WHEN NOT MATCHED - INSERT</strong><br>
  If the row exists in the source but not in the target...

  INSERT (...) VALUES (...) - A new row is inserted into the target table with values from the source.

  <br>
  <li><strong>Step 6: Final Result</strong><br>
  This single MERGE INTO command:
  <ul>
  <li>Updates rows when status = 'update'</li>
  <li>Deletes rows when status = 'delete'</li>
  <li>Inserts new rows not present in the target</li>
  </ul>
  <br>
  This pattern is perfect for incremental data loads, CDC, or SCD Type 1 patterns. Matt's row (id=4) was not in the source table, so it remains unchanged in the target.
